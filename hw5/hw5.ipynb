# 作业5

基于DTdata数据集，构建一颗决策树并可视化输出，并对下述记录进行测试：
Outlook=“sunny”, Temperature=“mild”, Humidity=“high”, Wind=FALSE

## 准备工作

### 加载包

```r
library(rpart)
library(rpart.plot)
```

### 加载数据

```r
weather <- read.csv("DTData.csv", sep=",")
# 将因变量设置为fact类型
weather$Play <- as.factor(weather$Play)
weather
```

## 数据处理

### 构建决策树

```r
# 需要设置一定的控制参数，否则会只剩下一个根节点
model <- rpart(Play ~ Outlook + Temperature + Humidity + Wind, data = weather, method="class", control = rpart.control(minsplit = 1,minbucket = 1,cp = 0))
rpart.plot(model)
```

### 预测

```r
x <- weather[4,]
# Outlook=“sunny”, Temperature=“mild”, Humidity=“high”, Wind=FALSE
x
```

```r
pred <- predict(model, newdata = x)
pred
```
