# 作业3

用sample函数产生XX名同学的成绩，范围为60到100之间，并计算最低分、最高分、平均分、中位数分数、四分位数、极差、平均差、方差、标准差、离散系数等统计指标。
**需同时列出函数和计算结果。**

获取值

```r
x <- 60:100
grades=sample(x,100,replace=TRUE)
grades
```

众数

```r
Mode <- function(data) {
    return(which.max(table(data)))
}
Mode(grades)
```

中位数

```r
median(x)
```

四分位数

```r
quantile(x)
```

均值

```r
mean(x)
```

调和平均数、几何平均数对此组数据无意义

极差

```r
range <- function(data){
    return (max(data)-min(data))
}
range(x)
```

平均差

```r
meandev <- function(data){
    s <- 0
    for (i in 1:length(data)){
        s = s + abs(data[i]-mean(data))
    }
    return (s / length(data))
}
meandev(x)
```

方差和标准差

```r
var(x)
sd(x)
```

离散系数

```r
vs <- function(data){
    return (sd(data) / mean(data))
}
vs(x)
```
