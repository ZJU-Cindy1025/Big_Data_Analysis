### 第一题

输出1至100之间的能够被3或5整除的数，并求和

```r
s <- 0
for (i in 1:100) {
    if (i %% 3 == 0 | i %% 5 == 0) {
        cat(i, " ")
        s <- s + i
    }
}
print(s)
```

### 第二题

输出1百万以内的斐波那契数列


```r
a1 <- 1
a2 <- 1
while (a1 <= 1000000) {
    cat(a1, " ")
    a3 <- a1 + a2
    a1 <- a2
    a2 <- a3
}
```
