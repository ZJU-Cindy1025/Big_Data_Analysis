# 作业9

利用R语言自带的women数据集，建立拟合身高和体重之间的关系的回归方程，画出拟合的线性回归线，并对回归方程进行检验。

## 加载数据

```r
data(women)
```

## 分析数据

### 建立线性回归模型

```r
model <- lm(women$height ~ women$weight)
model
```

```r
summary(model)
```

从上述结果中可以看到，体重$w$和身高$h$的关系是$w=0.28h+25.72$

另外，可以得到$R^2$的值是0.9903，说明回归模型可以解释99.03%的数据方差。另外，F检验的P检验结果为$1.09 \times 10^{-14}$，小于0.05；系数$b$的t检验结果为$2.68 \times 10^{-12}$，小于0.05；相关系数$r$的t检验结果为$1.09 \times 10^{-14}$，小于0.05，说明回归模型显著，得出的结论有效

### 图表绘制

```r
plot(women$weight, women$height, main = "体重和身高的线性回归", xlab = "身高", ylab = "体重")
abline(model)
```

从上述图中可以看到，体重和身高大致呈现线性正相关关系。
