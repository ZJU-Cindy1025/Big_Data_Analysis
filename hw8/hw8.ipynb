## 数据获取

### 获取影评

使用python爬虫脚本爬取影评

```python
%run 豆瓣电影获取.py
```

将结果转为字符串

```python
text = open("阿甘正传--影评.txt",encoding="utf8").read()
print(text)
```

### 导入停用词表

```python
# 停用词表
stop_words = open("stop_words_full.txt",encoding="utf8").read().split("\n")
```

## 数据分析

### 分割词汇

使用python的jieba库分割中文词汇和英文词汇

```python
import jieba
result = ' '.join(jieba.cut(text))
print(result)
```

### 生成词云图

设置停用词表，生成词云图

```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud
wc = WordCloud(font_path='msyh.ttc',background_color='white',stopwords=stop_words).generate(text)
```

使用matplotlib绘制词云图

```python
plt.imshow(wc, interpolation='bilinear')
plt.axis('off')
```
